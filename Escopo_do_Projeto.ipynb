1. Visão Geral

O sistema tem como objetivo disponibilizar uma interface web para consulta, filtro e exportação de registros de um banco de dados PostgreSQL.
O usuário poderá aplicar filtros, visualizar resultados e exportar em CSV ou PDF, garantindo uma experiência rápida e intuitiva.

2. Arquitetura de Componentes
Componente	Tecnologia	Descrição
Frontend	Vue.js	Interface de usuário com filtros, visualização e botões para exportação.
API Backend	C# - ASP.NET Core	Lógica de negócios, autenticação, integração com PostgreSQL e roteamento de requisições.
Serviço Exportação	Python	Responsável por gerar arquivos CSV/PDF conforme filtros aplicados pelo usuário.
Banco de Dados	PostgreSQL	Armazena os registros que serão consultados, filtrados e exportados.
